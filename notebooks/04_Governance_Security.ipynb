```python
%sql
CREATE OR REPLACE FUNCTION nyc_taxi_lakehouse.gold.mask_revenue(revenue DOUBLE)
RETURN IF(
    is_account_group_member('finance_team'),
    revenue,
    0.0
);

ALTER TABLE nyc_taxi_lakehouse.gold.payment_type_summary
ALTER COLUMN total_revenue SET MASK nyc_taxi_lakehouse.gold.mask_revenue;
```

```python
%sql
CREATE OR REPLACE FUNCTION nyc_taxi_lakehouse.gold.filter_payment_method(payment_method STRING)
RETURN IF(
    is_account_group_member('global_admins'),
    TRUE,
    payment_method = 'Credit Card'
);

ALTER TABLE nyc_taxi_lakehouse.gold.payment_type_summary
SET ROW FILTER nyc_taxi_lakehouse.gold.filter_payment_method ON (payment_method);
```
